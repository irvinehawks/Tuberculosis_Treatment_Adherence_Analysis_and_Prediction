```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

df = pd.read_csv('../data/raw/tb_adherence_practice_raw.csv')

# Save the df.head
findings_dir = Path('../data/findings')
findings_dir.mkdir(exist_ok=True)

# inspecting duplicates before removing them
duplicates = df[df.duplicated(keep=False)]
duplicates

# Save file for reports
duplicates.to_csv(
    findings_dir / "duplicate_records.csv" ,
    index=False
)

# Remove duplicates to start cleaning the data
df = df.drop_duplicates()

df.duplicated().sum()

# Handle missing values
df.isnull().sum()

df["monthly_income_usd"].describe()
df["monthly_income_usd"] = df["monthly_income_usd"].fillna(
    df["monthly_income_usd"].median()
)

df["side_effects"].describe()
df["side_effects"] = df["side_effects"].fillna(
    df["side_effects"].mode().iloc[0]
)

df["support_score_1_10"].describe()
df["support_score_1_10"] = df["support_score_1_10"].fillna(
    df["support_score_1_10"].median()
)

# Validate numerical ranges
df["age"].describe()
df["distance_to_clinic_km"].describe()
df["clinic_wait_time_min"].describe()
df["support_score_1_10"].describe()

df.to_csv('../data/processed/tb_adherence_cleaned.csv', index=False)
```
