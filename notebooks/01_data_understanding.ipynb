```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

df = pd.read_csv('../data/raw/tb_adherence_practice_raw.csv')

# Save the df.head
findings_dir = Path('../data/findings')
findings_dir.mkdir(exist_ok=True)
df.head().to_csv(findings_dir / 'df_head.csv', index=False)
df.head()

# Save the df.shape
with open(findings_dir / "df_shape.txt", "w") as f:
    f.write(f"Rows: {df.shape[0]}\n")
    f.write(f"Columns: {df.shape[1]}\n")
    f.write(f"Shape: {df.shape}\n")

df.shape

# Save the df.describe
df.describe().to_csv(
    findings_dir / 'df_describe.csv'
)

df.describe()

df.info()

# inspecting categorical variables/ data
df.describe(include = "object")
df["sex"].value_counts()
df["district"].value_counts()
df["employment_status"].value_counts()
df["side_effects"].value_counts()
df["adherent"].value_counts()

# Finding missing values
df.isnull().sum().to_csv(
    findings_dir / "missing_values.csv"
)

# Finding duplicate records
df.duplicated().sum()
df.duplicated().to_csv(
    findings_dir / "duplicate_records.csv"
)
```
